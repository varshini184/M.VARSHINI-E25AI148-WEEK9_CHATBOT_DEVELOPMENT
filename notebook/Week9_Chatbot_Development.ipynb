## 2. Implementing Conversation History

Conversation history allows the chatbot to remember previous messages during a conversation. The stored messages are sent to the language model along with each new question, helping it generate context-aware responses.

```python

# Initialize conversation history

conversation_history = [
    {
        "role": "system",
        "content": (
            "You are an education chatbot. "
            "Explain concepts in simple language "
            "and provide examples for students."
        )
    }
]

print("Conversation history initialized successfully.")
```

```python

# First question
question1 = "My name is Varshini."

conversation_history.append({
    "role": "user",
    "content": question1
})

response1 = ollama.chat(
    model=model_name,
    messages=conversation_history
)

answer1 = response1["message"]["content"]

conversation_history.append({
    "role": "assistant",
    "content": answer1
})

print("User:", question1)
print("Chatbot:", answer1)
```

```python

question2 = "What is my name?"

conversation_history.append({
    "role": "user",
    "content": question2
})

response2 = ollama.chat(
    model=model_name,
    messages=conversation_history
)

answer2 = response2["message"]["content"]

conversation_history.append({
    "role": "assistant",
    "content": answer2
})

print("User:", question2)
print("Chatbot:", answer2)
```

## 3. Saving Conversation History to CSV

The chatbot saves user questions, chatbot responses, and timestamps in CSV files for future analysis.

```python

import csv
import os
from datetime import datetime

# Create the dataset folder path
dataset_folder = os.path.abspath("../dataset")

# If the path does not exist, use the project dataset folder
if not os.path.isdir(dataset_folder):
    dataset_folder = os.path.join(os.getcwd(), "dataset")

os.makedirs(dataset_folder, exist_ok=True)

history_file = os.path.join(dataset_folder, "chat_history.csv")
responses_file = os.path.join(dataset_folder, "responses.csv")

# Create CSV files with column headings if needed
for file_path in [history_file, responses_file]:
    if not os.path.exists(file_path) or os.path.getsize(file_path) == 0:
        with open(file_path, "w", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)
            writer.writerow(
                ["Timestamp", "User_Question", "Chatbot_Response"]
            )

print("CSV file paths:")
print(history_file)
print(responses_file)
```

```python

# Save the two questions and responses from Step 13

records = [
    (question1, answer1),
    (question2, answer2)
]

timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

for question, answer in records:
    for file_path in [history_file, responses_file]:
        with open(file_path, "a", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)
            writer.writerow([timestamp, question, answer])

print("Conversation saved successfully!")
```

## 4. Testing the Chatbot

The chatbot is tested using factual, reasoning, coding, and general conversation questions to evaluate its responses across different tasks.

```python

import csv
from datetime import datetime

test_questions = [
    ("Factual", "What is artificial intelligence?"),
    ("Reasoning", "Why are loops useful in programming?"),
    ("Coding", "Write a Python program to add two numbers."),
    ("General", "Give me three tips to prepare for an examination.")
]

for category, question in test_questions:
    print("\nCategory:", category)
    print("User:", question)

    try:
        response = ollama.chat(
            model=model_name,
            messages=[
                {
                    "role": "system",
                    "content": (
                        "You are an education chatbot. "
                        "Give clear, simple and helpful answers."
                    )
                },
                {"role": "user", "content": question}
            ]
        )

        answer = response["message"]["content"]
        print("Chatbot:", answer)

        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        for file_path in [history_file, responses_file]:
            with open(
                file_path, "a", newline="", encoding="utf-8"
            ) as file:
                writer = csv.writer(file)
                writer.writerow([timestamp, question, answer])

    except Exception as error:
        print("Error:", error)

print("\nChatbot testing completed.")
```
